# Data Reading

```python
spark
```

```python
df=spark.read.format('parquet')\
        .load("abfss://bronze@carprojectazurestorage.dfs.core.windows.net/rawdata")
```

```python
df.display()
```

# Data Transformation


```python
from pyspark.sql.functions import *
df=df.withColumn('Category',split(col('Model_ID'),'-')[0])
df.display()
```

```python
df.printSchema()
```

```python
df.select("Category").show()
```

```python
df=df.withColumn('Revenue_per_unit',col('Revenue')/col('Units_Sold'))
df.display()
```

```python
df_agg=df.groupBy("Year","BranchName").agg(sum("Units_sold").alias("Total_units_sold")).sort("Year","Total_units_sold",ascending=[1,0])
df_agg.display()
```

```python
df.write.mode('overwrite')\
    .format('parquet')\
    .option("path",'abfss://silver@carprojectazurestorage.dfs.core.windows.net/carsales')\
    .save()
```
